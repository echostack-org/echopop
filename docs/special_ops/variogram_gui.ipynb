In some cases, the appropriate (semi)variogram model and its respective parameters are not known ahead of time. Furthermore, it may also be unclear how to appropriately bound each parameter for optimization like in [the example for the `echopop.geostatistics.Variogram` class](variogram_class). Echopop provides an alternative GUI-based approach for parameterizing the (semi)variogram. This GUI acts as a wrapper around the {py:func}`Variogram <echopop.geostatistics.Variogram>` class, so the underlying implementation for the empirical variogram estimation and model fitting are identical to when the `Variogram` class is used directly.

## Import necessary modules

The {py:class}`VariogramGUI <echopop.graphics.VariogramGUI>` class can be imported directly from the `echopop.graphics` sub-package. This is an interactive GUI that allows one to compute, visualize, and optimize empirical and theoretical variograms from spatial data, all within a Jupyter environment.

``````python
from echopop.graphics import VariogramGUI
``````

`````{admonition} Environment
:class: danger
The `echopop.graphics.VariogramGUI` class is built using interactive widgets and other modules from `ipywidgets` and related libraries using a `bokeh` graphics backend. Consequently, this backend requires this class to be exclusively used within a Jupyter notebook. It will not render correctly when ran in a non-interactive environment (e.g., terminal, console, script).
`````

## `VariogramGUI`

The `.VariogramGUI` class follows a similar convention for initializing the `Variogram` class. The arguments include:

- `data`: `The transect data.
- `lag_resolution`: The distance interval represented by each lag. This is a positive-only quantity that determins the spatial resolution of the variogram analysis.
- `n_lags`: The number of lags used for computing the semivariance.
- `coordinates`: The column names for the spatial coordinates in the transect dataset.
- `variogram_parameters`: An optional dictionary that contains initial values for certain variogram parameters (e.g., `"sill"`).

The `VariogramGUI` class object is created via:

``````python
gui = VariogramGUI(
    data=df_nasc_no_age1_prt,
    lag_resolution=0.002,
    n_lags=30,
    coordinates=("x", "y")
)
``````

This initializes the graphics backend and sets up the class to store updated parameter values when adjusting the variogram models or running the optimization. Just like in the `Variogram` class, the parameters can be inspected via the object's attributes. Specifically:

- `empirical_results`: The empirical variogram.
- `theoretical_results`: The initial variogram model parameters and resulting estimates at each lag.
- `optimization_results`: The optimized variogram model parameters. 

Since the object has not been run yet, these are all empty dictionaries.

## Rendering the GUI

The GUI can be rendered by calling the name of the object. The GUI comprises **5** tabs.

1. **Instructions**: This provides a complete set of instructions for how to use the GUI. This serves as the primary documentation for the GUI and its general usage.
2. **Initialize variogram**: This initializes the GUI. This is helpful for when one may want to change the lag resolution or number of lags without needing to close out the GUI and re-run it with new input values.
3. **Empirical variogram**: This computes the empirical variogram with an optional azimuth angle filter that helps to constrain anisotropy.
4. **Variogram model**: A dropdown selection menu provides a list of compatible models with the GUI. As each model is selected, the model parameters are dynamically populated. 
5. **Optimize variogram parameters**: Variogram model parameters can be fine-tuned (or held constant) for least-squares fitting. Different keyword arguments compatible with the `lmfit.minimize` function can also be provided in dictionary format that can adjust how the least-squares fitting is performed.

### Instructions
This provides a complete set of instructions for how to use the GUI. This serves as the primary documentation for the GUI and its general usage.

``````python
import echopop.graphics.variogram_gui as vargui
import ipywidgets as ipw
import panel as pn

inst_tab = vargui.instructions_tab(gui)
inst_plot = ipw.HBox(
    [
        inst_tab,
        ipw.VBox([gui.plot_output, gui.status_pane]),
    ],
    layout=ipw.Layout(width="100%"),
)
out = pn.panel(inst_plot)
out.save("../_static/vargui_instructions.html", embed=True, embed_json=True)
``````
